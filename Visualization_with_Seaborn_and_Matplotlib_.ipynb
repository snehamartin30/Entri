**Visualization with Seaborn and Matplotlib**

```python
#Importing all the essential libraries.
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

#Loading the Titanic dataset.
df = sns.load_dataset("titanic")

#previewing the dataset.
print(df.head())
```

```python
# 1. Countplot: Passenger Count by Embarked Port.
plt.figure(figsize = (7,5))
sns.countplot(data = df, x = "embarked", hue = "sex", palette = "bright")
plt.title("Passenger count by embarked port and gender",fontsize = 15, fontweight = "bold")
plt.xlabel("Embraked Port",fontsize = 12.5)
plt.ylabel("Passenger count",fontsize = 12.5)
plt.legend(title = "Gender")
plt.show()
print("Southampton (S) saw the highest volume of passengers embark overall for both males and females, followed by Cherbourg (C) and Queenstown (Q).")
```

```python
# 2.Swarmplot: Age Distribution across Classes.
plt.figure(figsize = (8,6))
sns.swarmplot(data = df , x = "pclass", y = "age", hue = "sex", dodge = True, palette = "bright", size = 4)
plt.title("Age Distribution Across Passenger Classes by Gender", fontsize = 15, fontweight = "bold")
plt.xlabel("passenger class (pclass)", fontsize = 12.5)
plt.ylabel("Age",fontsize = 12.5)
plt.legend(title = "Gender")
plt.show()
print("""OBSERVATION : First-class passengers were notably older overall, while third class had a much younger population base.
Within each class, male and female age spreads were fairly similar, though first class contained notable groups of older men and women.""")
```

```python
# 3. Pie Chart: Gender Distribution
plt.figure(figsize = (6,6))
# Count the number of male and female passengers.
gender_counts = df['sex'].value_counts()
#Explode the first slice (male).
explode = (0.09,0)
# Create the pie chart.
plt.pie(gender_counts.values,
        labels=gender_counts.index,
        autopct='%1.1f%%',
        explode=explode,
        colors=['green', 'yellow'],
        shadow=True)
plt.title('Gender Distribution of Passengers', fontsize=14, fontweight='bold')
plt.show()
print("""OBSERVATION : There were substantially more male passengers on board than female passengers.""")
```

```python
#4.Heatmap: Correlation Matrix.
plt.figure(figsize = (8,6))
# Selecting the specific columns.
corr_columns = ["age","fare","pclass","sibsp","parch","survived"]
corr_matrix = df[corr_columns].corr()
#Create heatmap
sns.heatmap(corr_matrix, annot = True, cmap = "viridis",)
plt.title('Correlation Matrix of Titanic Features', fontsize=14, fontweight='bold')
plt.show()
print("""OBSERVATION : The correlation matrix shows that fare has a positive correlation with survival and a strong negative correlation with pclass.
Age exhibits a slight negative correlation with survival. """)
```

```python
#5.Violin Plot: Fare by Class and Gender
plt.figure(figsize = (9,6))
# Creating violin plot.
sns.violinplot(x = "pclass",y = "fare", hue = "sex", data = df, palette = "bright")
plt.title("Fare Distribution Across Classes by Gender", fontsize=14, fontweight="bold")
plt.xlabel("Passenger Class (Pclass)", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.legend(title="Gender")
plt.show()
print("""OBSERVATION : The violin plot clearly shows that first-class fares had a massive spread with extreme far outliers,
whereas third-class fares were tightly clustered at very low price points with a small variation across genders.""")
```

```python
#6.Scatter Plot: Age vs Fare.
plt.figure(figsize=(9, 6))

# Create the scatter plot
sns.scatterplot(x='age', y='fare', data = df, hue='survived', palette='viridis', alpha=0.7)

plt.title('Relationship Between Age and Fare Colored by Survival Status', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Fare', fontsize=12)
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
print("""OBSERVATION : The scatter plot illustrates that higher-fare paying passengers experienced noticeably better survival rates across various age groups,
 while the majority of non-survivors were concentrated among lower-fare payers.""")
```
